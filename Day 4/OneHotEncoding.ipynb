# Fraud Oracle Analysis using One Hot Encoding


```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression

df = pd.read_csv('fraud_oracle.csv')

df = df.dropna()
df = df.drop_duplicates()

display(df.head())

dummies = pd.get_dummies(df.Make)

merged = pd.concat([df, dummies], axis='columns')

final = merged.drop(['Make', 'Vulnerable'], axis='columns', errors='ignore')


final = final.drop(['Accura'], axis='columns', errors='ignore')

X = final.select_dtypes(include=[np.number]).drop('FraudFound_P', axis='columns')
y = final.FraudFound_P

model = LogisticRegression(max_iter=1000)
model.fit(X, y)

print(f"Predictions: {model.predict(X)[:5]}")
print(f"Model Score: {model.score(X, y)}")
```

## One Hot Encoding using Sklearn LabelEncoder and OneHotEncoder

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer

le = LabelEncoder()
dfle = df.copy()

dfle.Make = le.fit_transform(dfle.Make)

X_sklearn = dfle[['Make', 'Age']].values
y_sklearn = dfle.FraudFound_P.values

ct = ColumnTransformer([('Make', OneHotEncoder(), [0])], remainder='passthrough')

X_transformed = ct.fit_transform(X_sklearn)

X_final = X_transformed[:, 1:]

model.fit(X_final, y_sklearn)
print(f"Accuracy with OneHotEncoder: {model.score(X_final, y_sklearn)}")
```
